```python
import py5
import py5_tools
import IPython.display as display
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
from PIL import Image
import numpy as np
import os

# Función helper para mostrar imágenes
def mostrar_imagen(ruta, titulo="", figsize=(8, 6)):
    """Muestra una imagen con matplotlib"""
    if os.path.exists(ruta):
        plt.figure(figsize=figsize)
        img = mpimg.imread(ruta)
        plt.imshow(img)
        plt.axis('off')
        if titulo:
            plt.title(titulo)
        plt.tight_layout()
        plt.show()
    else:
        print(f"No se encontró la imagen: {ruta}")

def aplicar_efectos_imagen(imagen_path, titulo_base="Imagen"):
    """Aplica diferentes efectos a una imagen y los muestra"""
    from PIL import Image, ImageEnhance
    
    if not os.path.exists(imagen_path):
        print(f"No se encontró la imagen: {imagen_path}")
        return
    
    # Cargar imagen con PIL
    img = Image.open(imagen_path)
    
    # Crear subplot para mostrar múltiples efectos
    fig, axes = plt.subplots(2, 2, figsize=(12, 10))
    fig.suptitle(f'{titulo_base} - Diferentes Efectos', fontsize=16)
    
    # Imagen original
    axes[0,0].imshow(img)
    axes[0,0].set_title('Original')
    axes[0,0].axis('off')
    
    # Efecto de alto contraste (THRESHOLD simulado)
    enhancer = ImageEnhance.Contrast(img)
    img_contrast = enhancer.enhance(2.0)
    axes[0,1].imshow(img_contrast)
    axes[0,1].set_title('Alto Contraste')
    axes[0,1].axis('off')
    
    # Efecto blanco y negro
    img_gray = img.convert('L')
    axes[1,0].imshow(img_gray, cmap='gray')
    axes[1,0].set_title('Blanco y Negro')
    axes[1,0].axis('off')
    
    # Efecto invertido
    img_array = np.array(img)
    img_inverted = 255 - img_array
    axes[1,1].imshow(img_inverted)
    axes[1,1].set_title('Colores Invertidos')
    axes[1,1].axis('off')
    
    plt.tight_layout()
    plt.show()

print("Librerías importadas y funciones helper definidas")
```

# Ejercicio 1: Figura en Espacio de Color RGB

1. Abran un nuevo sketch en py5.
2. Establezcan un lienzo (canvas) de 400x400 píxeles.
3. Asegúrense de que el modo de color esté configurado en RGB.
4. Dibujen una figura geométrica simple.
5. Asignen un color de relleno a la figura utilizando valores RGB.
6. Guarden la imagen resultante en formato JPEG con el nombre figura_rgb.jpg.

```python
import py5

def setup():
    # Establecer el tamaño del lienzo en 400x400 píxeles
    py5.size(400, 400)
    py5.color_mode(py5.RGB, 255)
    py5.no_stroke()

    # Asignar un color de relleno
    py5.fill(255, 0, 255)

    # Dibujar una figura geométrica simple: un triángulo
    py5.triangle(200, 100, 100, 300, 300, 300)

    # Guardar la imagen
    py5.save_frame("figura_rgb.jpg")

    # Tomar una captura de pantalla
    py5_tools.screenshot()

    # Salir del sketch después de guardar la imagen
    py5.exit_sketch()

py5.run_sketch()
print("Figura RGB creada y guardada como figura_rgb.jpg")
```

```python
# Mostrar la figura RGB generada
mostrar_imagen('figura_rgb.jpg', 'Figura en Espacio de Color RGB', (6, 6))
```

# Construcción y Uso de la Cámara Oscura

En esta etapa, pasarán de lo digital a lo analógico para entender el proceso fundamental de la formación de imágenes.

## Materiales:

* Una caja de zapatos con tapa.
* Cinta adhesiva opaca (cinta aisladora negra es ideal).
* Papel manteca o de calcar.
* Un alfiler o aguja.
* Tijera o cúter.
* Un teléfono celular con cámara.

## Instrucciones:

**Construcción de la Cámara:**
- En uno de los extremos más cortos de la caja, realicen con el alfiler un orificio muy pequeño y prolijo en el centro. Este será su estenopo (pinhole).
- En el extremo opuesto, recorten una ventana rectangular, dejando un marco de aproximadamente 2 cm.
- Peguen con la cinta un trozo de papel manteca sobre esta ventana, por la parte interior de la caja.
- Coloquen la tapa y sellen con cinta cualquier otra entrada de luz.

**Captura de la Imagen:**
- Dirijan el estenopo hacia una fuente de luz intensa.
- Observen la imagen que se proyecta invertida sobre la pantalla de papel manteca.
- Utilizando la cámara de su teléfono, tomen la mejor fotografía posible.

## Imagen en diferentes tintes

Aplicando diferentes filtros de color usando py5 para observar cómo afectan la imagen capturada.

```python
def mostrar_tintes():
    """Muestra la imagen con diferentes tintes de color"""
    img_color = py5.load_image('002.jpeg')
    
    py5.size(800, 200)
    py5.color_mode(py5.HSB, 360, 100, 100)
    py5.no_stroke()

    # Imagen original
    py5.no_tint()
    py5.image(img_color, 0, 0, 200, 200)

    # Tinte Rojo
    py5.tint(150, 150, 150)
    py5.image(img_color, 200, 0, 200, 200)

    # Tinte Verde
    py5.tint(50, 50, 50)
    py5.image(img_color, 400, 0, 200, 200)

    # Tinte Azul
    py5.tint(10, 10, 10)
    py5.image(img_color, 600, 0, 200, 200)

    py5.save_frame("imagen_tintes.jpg")
    py5.exit_sketch()

py5.run_sketch()
print("Efectos de tintes aplicados y guardados")
```

## Imagen con alto contraste

Ajustamos a 0.6 THRESHOLD donde se capta una mejor visualización.

```python
def aplicar_contraste():
    """Aplica filtro de alto contraste"""
    py5.size(400, 400)
    
    img = py5.load_image("002.jpeg")
    py5.image(img, 0, 0, 400, 400)
    
    # Aplicar filtro de umbral (simula THRESHOLD)
    py5.apply_filter(py5.THRESHOLD, 0.6)
    
    py5.save_frame("imagen_contraste.jpg")
    py5.exit_sketch()

py5.run_sketch()
print("Filtro de alto contraste aplicado")
```

## Imagen en Blanco y Negro

```python
def aplicar_blanco_negro():
    """Aplica filtro en escala de grises"""
    py5.size(400, 400)
    
    img = py5.load_image("002.jpeg")
    py5.image(img, 0, 0, 400, 400)
    
    # Aplicar filtro de escala de grises
    py5.apply_filter(py5.GRAY)
    
    py5.save_frame("imagen_byn.jpg")
    py5.exit_sketch()

py5.run_sketch()
print("Filtro en escala de grises aplicado")
```

## Imagen con colores Invertidos

```python
def aplicar_invertido():
    """Aplica filtro de inversión de colores"""
    py5.size(400, 400)
    
    img = py5.load_image("002.jpeg")
    py5.image(img, 0, 0, 400, 400)
    
    # Aplicar filtro de inversión
    py5.apply_filter(py5.INVERT)
    
    py5.save_frame("imagen_invertida.jpg")
    py5.exit_sketch()

py5.run_sketch()
print("Filtro de inversión de colores aplicado")
```

## Observaciones:

Con los resultados obtenidos encontramos que el algoritmo detecta perfectamente los colores de la imagen capturada.

* **En diferentes tonos**: La imagen es visible pese a estar no tan nítida
* **Más contraste**: Mejora la separación entre luces y sombras; se perciben más la forma del árbol.
* **Blanco y Negro**: Todo parece más suave, con menos definición y profundidad.

Es interesante notar que el aumento de 'nitidez' logrado por contraste no requiere mejorar la resolución, sino realzar contornos. Es fundamental equilibrar estos ajustes para mantener la fidelidad y autenticidad visual de la imagen.
